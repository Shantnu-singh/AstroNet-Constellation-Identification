```python
import copy

original = [[1, 2, 3], [4, 5, 6]]
shallow = copy.copy(original)

shallow[0][0] = 99  # Changes reflect in both

print(original)  # [[99, 2, 3], [4, 5, 6]]
print(shallow)   # [[99, 2, 3], [4, 5, 6]]
```
