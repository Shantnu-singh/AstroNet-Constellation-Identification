```python
def swap(lst , i , j):
    lst[i] , lst[j] = lst[j] , lst[i]
    
def heplify(arr , n , i):
    
    largest = i # root is largest
    
    left_node = 2*i+1
    right_node = 2*i+2
    
    if left_node < n and arr[left_node] > arr[largest] :
        largest = left_node
        
    if  right_node < n and arr[right_node] > arr[largest]:
        largest = right_node
    
    if largest != i:
        swap(arr , largest , i)
        
        heplify(arr , n , largest)
    
def build_heap(arr , n):
    startIdx = n//2 - 1
    
    for i in range(startIdx , -1 , -1):
        heplify(arr , n , i)
    
    return arr
        
arr = [1, 3, 5, 4, 6, 13,
           10, 9, 8, 15, 17]

build_heap(arr , len(arr))
arr
    
```

```python
# Heap Sort

def heapify(arr , n , i):
    
    largest = i
    
    left_node = 2*i +1 
    right_node = 2*i + 2
    
    if left_node < n and arr[left_node] > arr[largest]:
        largest = left_node
        
    if right_node < n and arr[right_node] > arr[largest]:
        largest = right_node
        
    if largest != i:
        swap(arr , i , largest)
        
        heapify(arr , n , largest)

def heap_sort(arr , n):
    
    # Build the heap
    for i in range(n//2-1 , -1 , -1):
         heapify(arr , n , i)
         
    
    # one by one remove the element
    for i in range(n-1 , 0 , -1):
         swap(arr , i , 0)
         heapify(arr , i , 0)
         
    return arr
arr = [1, 3, 5, 4, 6, 13,
           10, 9, 8, 15, 17]

heap_sort(arr , len(arr))
```
