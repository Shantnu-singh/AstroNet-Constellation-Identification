```python
def swap(lst , i , j):
    lst[i] , lst[j] = lst[j] , lst[i]
    
def heplify(arr , n , i):
    
    largest = i # root is largest
    
    left_node = 2*i+1
    right_node = 2*i+2
    
    if left_node < n and arr[left_node] > arr[largest] :
        largest = left_node
        
    if  right_node < n and arr[right_node] > arr[largest]:
        largest = right_node
    
    if largest != i:
        swap(arr , largest , i)
        
        heplify(arr , n , largest)
    
def build_heap(arr , n):
    startIdx = n//2 - 1
    
    for i in range(startIdx , -1 , -1):
        heplify(arr , n , i)
    
    return arr
        
arr = [1, 3, 5, 4, 6, 13,
           10, 9, 8, 15, 17]

build_heap(arr , len(arr))
arr
    
```

```python
# Heap Sort

def heapify(arr , n , i):
    
    largest = i
    
    left_node = 2*i +1 
    right_node = 2*i + 2
    
    if left_node < n and arr[left_node] > arr[largest]:
        largest = left_node
        
    if right_node < n and arr[right_node] > arr[largest]:
        largest = right_node
        
    if largest != i:
        swap(arr , i , largest)
        
        heapify(arr , n , largest)

def heap_sort(arr , n):
    
    # Build the heap
    for i in range(n//2-1 , -1 , -1):
         heapify(arr , n , i)
         
    
    # one by one remove the element
    for i in range(n-1 , 0 , -1):
         swap(arr , i , 0)
         heapify(arr , i , 0)
         
    return arr
arr = [1, 3, 5, 4, 6, 13,
           10, 9, 8, 15, 17]

heap_sort(arr , len(arr))
```

```python
# Insertion Sort
def selection_sort(arr ,n):
    for i in range(n):
        min = i
        for j in range(i+1 , n):
            if arr[j] < arr[min]:
                min = j
        
        swap(arr, i , min)
        
arr = [1, 3, 5, 4, 6, 13,
           10, 9, 8, 15, 17]

selection_sort(arr , len(arr))
arr
```

```python
# Insertion Sort

def insertion_sort(arr , n):
    for i in range(1 , n):
        min_index = i
        j = i-1
        while(j >= 0 and arr[min_index] < arr[j]):
            arr[j+1] = arr[j]
            j -= 1
            
    return arr

arr = [1, 3, 5, 4, 6, 13,
        10, 9, 8, 15, 17]

insertion_sort(arr , len(arr))
arr        
            
            
```

```python
# Merge sort

def merge(arr , left , right , mid):
    i = left
    j = mid+1
    temp = []
    
    while(i <= mid and j <= right):
        
        while(arr[i] < arr[j]):
            temp.append(arr[i])
            i += 1
        while( arr[j] < arr[i] ):
            temp.append(arr[j])
            j +=1
            
    while(i <= mid):
        temp.append(arr[i])
        i += 1
    while(j <= right):
        temp.append(arr[j])
        j += 1
        
    arr = temp
    

def merge_sort(arr , left ,right):
    if left < right:
        mid = (left+right) // 2
        
        merge_sort(arr , left, mid)
        merge_sort(arr , mid+1 , right)
        merge(arr , left , right , mid)
        

arr = [1, 3, 5, 4, 6, 13,
        10, 9, 8, 15, 17]

merge_sort(arr , left= 0 , right= len(arr)-1)
arr        
```

```python
# Quick sort --> divide and conqure

def partition(arr , low , high):
    pivot = arr[low]
    
    i = low
    j = high
    
    while(i < j):
        while i < high and arr[i] <= pivot :
            i+=1
        while j > low and arr[j] > pivot and 1:
            j -= 1
        if (i < j) :
            swap( arr, i , j)
    
    swap(arr , low , j )
    return j
    
    


def quick_sort(arr , low , high):
    
    if low < high:
        
        par = partition(arr , low , high)
        
        # recursion
        quick_sort(arr , low , par-1)
        
        quick_sort( arr, par+1 , high )
        
arr = [1, 3, 5, 4, 6, 13,
        10, 9, 8, 15, 17]

quick_sort(arr , low= 0 , high= len(arr)-1)
arr       
```

```python
# Parent class
class Animal:
    def __init__(self):
        print("this is base class") # Initialize the name attribute

    def speak(self):
        pass  # Placeholder method to be overridden by child classes

# Child class inheriting from Animal
class Dog(Animal):
    def __init__(self):
        print("child")
    def speak(self):
        return f" barks!, {super().__init__()}"  # Override the speak method

# Creating an instance of Dog
dog = Dog()
print(dog.speak())  # Output: Buddy says Woof!
```
